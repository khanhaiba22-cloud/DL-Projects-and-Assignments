```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def step_function(x):
    return np.where(x >= 1, 1, 0)
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([0, 1, 1, 1])

weights = np.array([0.0,0.0])
learning_rate = 0.5
epochs = 10


for epoch in range(epochs):  
    print(f"\nEpoch {epoch + 1}")
    errors = 0
    
    for i in range(len(X)):  
        linear_output = np.dot(X[i], weights)
        prediction = step_function(linear_output)
        
        error = y[i] - prediction
        
        if error != 0:
            errors += 1
            weights += learning_rate * error * X[i]
            
        print(f"Input: {X[i]}, Target: {y[i]}, Prediction: {prediction}, Weights: {weights} ")   
            

    if errors == 0:
        print("\nTraining completed successfully!")
        break
    
    
print("\nfinal weights: ",weights)

            

```

```python
plt.figure(figsize=(6,6))


for i in range(len(X)):
    if y[i] == 0:
        plt.scatter(X[i,0], X[i,1],
                    color='red',
                    marker='o',
                    s=120,
                    label='Class 0' if i == 0 else "")
    else:
        plt.scatter(X[i,0], X[i,1],
                    color='blue',
                    marker='^',
                    s=120,
                    label='Class 1' if i == 1 else "")

x_values = np.linspace(-0.2, 1.2, 100)

if weights[1] != 0:
    y_values = (1 - weights[0] * x_values) / weights[1]
    plt.plot(x_values, y_values,
             color='black',
             linewidth=2,
             label='Decision Boundary')

plt.xlim(-0.2, 1.2)
plt.ylim(-0.2, 1.2)

plt.xlabel("Input x1")
plt.ylabel("Input x2")
plt.title("Single Layer Perceptron Decision Boundary")
plt.grid(True)
plt.legend()

plt.show()
```
